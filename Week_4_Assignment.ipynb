 **Exploring the Mathematics of the Data**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

df = pd.read_csv("/content/kenya_petroleum_stations_safety_dataset-3.csv")
print("Dataset shape", df.shape)
df.head(5)
```

```python
#Selecting 4 numerical variables relevant to the project
selected_variables =["num_dispensers", "num_staff", "pms_storage_capacity_litres", "ago_storage_capacity_litres"]

#Calculating the mean, median, range, variance, and standard deviation using Python
statistics = pd.DataFrame(index=selected_variables)
statistics['Mean'] = df[selected_variables].mean()
statistics['Median'] = df[selected_variables].median()
statistics['Range'] = df[selected_variables].max() - df[selected_variables].min()
statistics['Variance'] = df[selected_variables].var(ddof=1)
statistics['Standard_Deviation'] = df[selected_variables].std(ddof=1)
display(statistics.round(2))
```

```python
#Covariance and correlation
x = df["num_dispensers"]
y = df["num_staff"]

covariance = x.cov(y)       # sample covariance
correlation = x.corr(y)    # Pearson correlation

print(f'Covariance: {covariance:.4f}')
print(f'Pearson correlation: {correlation:.4f}')
```

```python
#Simple plot showing the relationship between the two variables.
x = df["num_dispensers"]
y = df["num_staff"]
plt.figure(figsize=(7, 5))
plt.scatter(x, y, alpha=0.7)
plt.title("num_dispensers vs num_staff")
plt.xlabel('num_dispensers')
plt.ylabel('num_staff')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

##What do the statistics tell about the data?

- **What do the statistics tell us?** The mean and median describe typical values. The range shows the spread between the minimum and maximum. Variance and standard deviation describe how widely observations vary; standard deviation is expressed in the original units, while variance is in squared units. Large differences between the mean and median can indicate skew or unusually high/low values.
- **What does the correlation tell us?** The calculated Pearson correlation summarizes the direction and strength of a *linear* association between num_dispensers and num_staff. A positive value means higher dispensers counts tend to occur with higher staff counts; a negative value means they tend to move in opposite directions; a value near zero indicates little linear association. For this dataset the two variables have a negative correlation.
- **Does correlation mean causation?** No. Correlation does not necessarily mean that one variable causes the other. Both variables may be influenced by other factors, or their association may occur by chance. Additional evidence and appropriate research methods would be needed to establish causation.
