#  Proposed ML workflow
**Data Collection**(Gather petroleum station information such as fuel storage capacity, fire equipment, leakage, electrical safety, environmental compliance, and inspection history) → **Data Cleaning**(Check for missing values, duplicates, inconsistent formats, and incorrect data) → **Exploratory Data Analysis**(Analyze the data to identify patterns and relationships between station characteristics and safety-risk levels) → **Feature Selection**(Select relevant input variables and convert categorical information such as Compliant/Partial/Non-compliant into numerical representations that ML models can process) → **Data Preprocessing**(Transforming and organising raw data into a structured format) → **Train/Test Split**(Divide the dataset into training data and testing data. The training data is used to build the model, while the testing data evaluates its performance) → **Model Training**(Train classification algorithms such as Logistic Regression and Random Forest to learn patterns associated with different risk levels) → **Model Evaluation**(Compare model predictions with actual risk classifications using accuracy, precision, recall, and a confusion matrix) → **Risk/Fire Prediction**(Use the best-performing model to classify a petroleum station as Low, Medium, High, or Critical risk) → **Prioritize High-Risk Stations**(The predicted risk can help prioritize stations requiring further inspection or safety intervention. The model supports, rather than replaces, professional safety inspections)
